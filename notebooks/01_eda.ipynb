```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8")
```

```python
labels_path = "../data/labels/deam_song_level_labels.csv"
df = pd.read_csv(labels_path)

df.head()
```

```python
print("Number of songs:", len(df))
print("\nColumns:")
df.columns
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(df["valence_mean"], bins=30)
plt.title("Valence Distribution")
plt.xlabel("Valence")

plt.subplot(1,2,2)
plt.hist(df["arousal_mean"], bins=30)
plt.title("Arousal Distribution")
plt.xlabel("Arousal")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df["valence_mean"], df["arousal_mean"], alpha=0.5)
plt.xlabel("Valence")
plt.ylabel("Arousal")
plt.title("Valence vs Arousal")
plt.grid(True)
plt.show()
```

### EDA Observations
- Valence and arousal values are continuous
- Distribution is reasonably balanced
- Weak to moderate correlation between valence and arousal
- Regression is appropriate for this task
